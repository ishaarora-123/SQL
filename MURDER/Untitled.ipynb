```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("murder.db")
```

```python
crime_scene_report = pd.read_csv("Data/crime scene report.csv")
drivers_license = pd.read_csv("Data/drivers license.csv")
fb_event_checkin = pd.read_csv("Data/fb event checkin.csv")
get_fit_now_checkin = pd.read_csv("Data/get fit now checkin.csv")
get_fit_now_member = pd.read_csv("Data/get fit now member.csv")
income = pd.read_csv("Data/income.csv")
interview = pd.read_csv("Data/interview.csv")
person = pd.read_csv("Data/person.csv")
```

```python
crime_scene_report.to_sql("crime_scene_report", conn, if_exists="replace", index=False)
drivers_license.to_sql("drivers_license", conn, if_exists="replace", index=False)
fb_event_checkin.to_sql("fb_event_checkin", conn, if_exists="replace", index=False)
get_fit_now_checkin.to_sql("get_fit_now_checkin", conn, if_exists="replace", index=False)
get_fit_now_member.to_sql("get_fit_now_member", conn, if_exists="replace", index=False)
income.to_sql("income", conn, if_exists="replace", index=False)
interview.to_sql("interview", conn, if_exists="replace", index=False)
person.to_sql("person", conn, if_exists="replace", index=False)
```
